# ICA Applied to Audio Unmixing

This notebook demonstrates Independent Component Analysis (ICA) for the blind source separation of audio mixtures. We apply a pipeline of `Centering -> Whitening -> ICA` to disentangle original sources from recorded audio mixtures.

Example dataset: `audio/run1/`

## 1. Import libs

```python
from pathlib import Path

import pandas as pd
import torch
from IPython.display import HTML, Audio, display

from ica.data.audio import AudioDataset
from ica.model import Kurtosis, Negentropy
from ica.preprocessing import Centering, Whitening
from ica.utils import Pipeline

_ = torch.manual_seed(0)


def display_audios(tensors, sample_rate, label):
    """Show one Audio player per tensor, side by side instead of stacked."""
    players = "".join(
        f'<div style="display:inline-block; margin-right:2em; text-align:center;">'
        f"<div>{label} {i}</div>{Audio(t.numpy(), rate=sample_rate)._repr_html_()}</div>"
        for i, t in enumerate(tensors)
    )
    display(HTML(players))


def side_by_side(*named_items):
    """Show labeled matrices (or anything with a pandas repr) next to each other."""
    blocks = "".join(
        f'<div style="display:inline-block; margin-right:2em; vertical-align:top;">'
        f"<b>{name}</b>{pd.DataFrame(matrix.numpy()).to_html()}</div>"
        for name, matrix in named_items
    )
    display(HTML(blocks))
```

## 2. Load and play wav files

```python
DATASET_DIR = Path("data/")
RUN_NAME = "run1"

dataset = AudioDataset(source=DATASET_DIR, alias=RUN_NAME)
dataset
```

```python
display_audios(dataset.to_tensor(), dataset.sample_rate, "mixture")
```

## 3. Data preparation: Centering + Whitening

Centering and whitening only need to be computed once. We build a preprocessing-only pipeline (no ICA model) and reuse the resulting whitened checkpoint `Z` for every method compared below, instead of rebuilding the pipeline per method.

```python
centering = Centering()
whitening = Whitening()

preprocessing = Pipeline(steps=[centering, whitening])
Z = preprocessing.fit_transform(dataset)
Z.shape
```

## 4. Apply every ICA method to the same prepared data

Each method only needs to `fit_transform` the checkpoint `Z` directly — preprocessing is not repeated per method.

```python
methods = {
    "Kurtosis": Kurtosis(),
    "Negentropy": Negentropy(),
}

results = {}
for name, model in methods.items():
    torch.manual_seed(0)
    results[name] = model.fit_transform(Z)

{name: Y.shape for name, Y in results.items()}
```

## 5. Recovered mixing matrices

After whitening, $z = V x_c$, where $V$ is the whitening matrix and $x_c$ are the centered mixtures. Each ICA method estimates an orthogonal unmixing matrix $W$ such that $y = Wz$. Solving for $x_c$ gives $x_c = V^{-1} W^\top y$, so $A = V^{-1} W^\top$ is the estimated mixing matrix (Sections 7.4, 8.4) — one per method, all built from the same $V$.

```python
def mixing_matrix(model):
    return torch.linalg.inv(whitening.whitening_matrix) @ model.components.T


mixing_matrices = {name: mixing_matrix(model) for name, model in methods.items()}
mixing_matrices
```

## 6. Separated sources

```python
for name, Y in results.items():
    display(HTML(f"<h4>{name}</h4>"))
    # ICA leaves both scale and sign undetermined; normalize to [-1, 1] before playback.
    Y_norm = Y / Y.abs().amax(dim=1, keepdim=True)
    display_audios(Y_norm, dataset.sample_rate, "source")
```

## 7. Comparing methods

The two mixing matrices, side by side.

```python
side_by_side(*mixing_matrices.items())
```
